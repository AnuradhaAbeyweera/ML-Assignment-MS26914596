```python
# Q-Learning Implementation
# import libraries
import numpy as np
import random
import matplotlib.pyplot as plt
```

```python
states = [0, 1]  # 0 = benign, 1 = malignant
actions = [0, 1] # predict benign / malignant

Q = np.zeros((2, 2))
```

```python
# Q-Learning Parameters and Training
alpha = 0.1
gamma = 0.9
episodes = 1000
rewards = []

for _ in range(episodes):
    state = random.choice(states)
    action = random.choice(actions)

    reward = 1 if action == state else -1
    rewards.append(reward)

    Q[state, action] = Q[state, action] + alpha * (
        reward + gamma * np.max(Q[state]) - Q[state, action]
    )
```

```python

# Plot Learning Progress
plt.plot(np.cumsum(rewards))
plt.title("Cumulative Reward Over Time")
plt.xlabel("Episodes")
plt.ylabel("Cumulative Reward")
plt.show()
```

```python
# Evaluate Final
correct = 0
total = 1000

for _ in range(total):
    state = random.choice(states)
    action = np.argmax(Q[state])

    if action == state:
        correct += 1

accuracy = correct / total
print("RL Accuracy:", accuracy)
```

```python
# print Q table
print("Q-Table:")
print(Q)

policy = np.argmax(Q, axis=1)
print("Policy:", policy)
```
