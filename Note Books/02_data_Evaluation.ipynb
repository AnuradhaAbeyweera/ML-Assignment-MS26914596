## Import Libraries

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

## Add Column Names and Load Dataset

```python
# Column names from wdbc.names
columns = [
    'ID', 'Diagnosis', 'radius_mean', 'texture_mean', 'perimeter_mean', 'area_mean', 'smoothness_mean',
    'compactness_mean', 'concavity_mean', 'concave_points_mean', 'symmetry_mean', 'fractal_dimension_mean',
    'radius_se', 'texture_se', 'perimeter_se', 'area_se', 'smoothness_se', 'compactness_se', 'concavity_se',
    'concave_points_se', 'symmetry_se', 'fractal_dimension_se',
    'radius_worst', 'texture_worst', 'perimeter_worst', 'area_worst', 'smoothness_worst',
    'compactness_worst', 'concavity_worst', 'concave_points_worst', 'symmetry_worst', 'fractal_dimension_worst'
]

# Load dataset
df = pd.read_csv('data/wdbc.data', names=columns)

# Quick look at data
df.head()
```

# Basic Dataset Exploration

```python
print("Shape:", df.shape)
df.info()
df.describe()
```

## Check Target Distribution

```python
df['Diagnosis'].value_counts()
```

```python
sns.countplot(x='Diagnosis', data=df)
plt.title("Distribution of Diagnosis (M vs B)")
plt.show()
```

## Convert Target Variable

```python
df['Diagnosis'] = df['Diagnosis'].map({'M': 0, 'B': 1})

df.head()
```

```python

# Separate data by class
malignant = df[df['Diagnosis'] == 0]
benign = df[df['Diagnosis'] == 1]
```

## Check Missing Values

```python
df.isnull().sum()
```

## Explore Feature Distributions

```python

features_to_plot = [
    'radius_mean',
    'texture_mean',
    'perimeter_mean',
    'area_mean'
]
plt.figure(figsize=(15, 20))

for i, feature in enumerate(features_to_plot, 1):
    plt.subplot(4, 2, 2*i - 1 )
    sns.histplot(df[feature], kde=True)
    plt.title(f'{feature} Distribution (Histogram + KDE)')

for i, feature in enumerate(features_to_plot, 1):
    plt.subplot(4, 2, 2*i)
    sns.boxplot(x=df[feature])
    plt.title(f'{feature} Distribution (Boxplot)')
```

```python
plt.figure(figsize=(15, 20))

for i, feature in enumerate(features_to_plot, 1):
     plt.subplot(4, 2, 2*i - 1 )
     sns.kdeplot(malignant[feature], label='Malignant', fill=True)
     sns.kdeplot(benign[feature], label='Benign', fill=True)

     plt.title(f"{feature} Distribution")
     plt.legend()
     #plt.show()

for i, feature in enumerate(features_to_plot, 1):
     plt.subplot(4, 2, 2*i )
     sns.boxplot(x='Diagnosis', y=feature, data=df)

     plt.title(f"{feature} vs Diagnosis")
     #plt.show()

```

## Analyze Feature Correlation

```python
# Compute correlation matrix
corr_matrix = df.corr()

# Plot heatmap
plt.figure(figsize=(20, 20))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f")

plt.title("Correlation Matrix Heatmap")
plt.show()
```

```python
selected_features = [
    'radius_mean',
    'texture_mean',
    'perimeter_mean',
    'area_mean',
    'Diagnosis'
]

sns.pairplot(df[selected_features], hue='Diagnosis')
plt.show()
```

```python
df = df.drop('ID', axis=1)
```
